# Repeats for Free Growing comparison $\Lambda = 1$ and $\Lambda = 5$, 8 repeats each

```python
repeats_lambda1 = pfunc.repeat_files(data_loc+"FreeGrow\\Lambda1",5)
repeats_lambda5 = pfunc.repeat_files(data_loc+"FreeGrow\\Lambda5",5)
repeats_lambda = repeats_lambda1 + pfunc.repeat_files(data_loc+"FreeGrow\\stress\\Lambda1",3) + repeats_lambda5 + pfunc.repeat_files(data_loc+"FreeGrow\\stress\\Lambda5",3)
Aplot.plot_growth_rate(repeats_lambda)
```

Run a student t-test to see the probability that the doubling times have the same expected values. In this test we assume that the two different sets of doubling times have the same variance. 

```python
tdoub_Lambda1_linear = [[0.722,0.742,0.752,0.721,0.725,0.723,0.783,0.757],[0.741,0.731,0.742,0.717,0.699]]
tdoub_Lambda5_linear = [[0.742,0.737,0.759,0.755,0.793,0.752,0.735,0.737],[0.767,0.74,0.771,0.754,0.804,0.761]]

tdoub_Lambda1_exp = [0.734,0.736,0.741,0.724,0.711,0.716]
tdoub_Lambda5_exp = [0.757,0.739,0.763,0.751,0.788,0.754]

results = ttest_ind(tdoub_Lambda1_linear[0],tdoub_Lambda5_linear[0]) #assuming they have same variation
print(f'the results for the t-test are: t-statistic = {round(results[0],4)}, pval = {round(results[1],4)}')
```
