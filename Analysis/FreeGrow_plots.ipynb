Run a Student t-test to see the probability that the doubling times have the same expected values. In this test, we assume that the two different sets of doubling times have the same variance. 

```python
tdoub_Lambda1_linear = [0.722,0.742,0.752,0.721,0.725,0.723,0.77,0.753]
tdoub_Lambda5_linear = [0.743,0.737,0.762,0.793,0.751,0.727,0.752,0.744,0.663]

tdoub_Lambda1_exp = [0.734,0.736,0.741,0.724,0.711,0.716]
tdoub_Lambda5_exp = [0.757,0.739,0.763,0.751,0.788,0.754]

results = ttest_ind(tdoub_Lambda1_linear,tdoub_Lambda5_linear) #assuming they have same variation
print(f'the results for the t-test are: t-statistic = {round(results[0],4)}, pval = {round(results[1],4)}')
```
